```python
def load_txt(path, label, src, category="unknown"):
    with open(path, encoding="utf-8") as f:
        lines = [l.strip() for l in f.readlines() if l.strip()]
    # strip leading "1.  " / "23. " numbering if present
    lines = [re.sub(r"^\d+\.\s*", "", l) for l in lines]
    return pd.DataFrame({
        "text": lines,
        "label": label,
        "category": category,
        "source": src,
        "source_url": np.nan
    })
```

```python
# Filling placeholders

PLACEHOLDER_MAP = {
    r"\[Company\]": "Acme Corp",
    r"\[Name\]": "John",
    r"\[Title\]": "Mr.",
    r"\[Date\]": "Monday",
    r"\[Time\]": "3 PM",
    r"\[Product\]": "your order",
    r"\[Number\]": "1234",
    r"\[City\]": "Boston",
}
def fill_placeholders(t):
    for pattern, repl in PLACEHOLDER_MAP.items():
        t = re.sub(pattern, repl, t)
    return t

df["text"] = df["text"].apply(fill_placeholders)
```

```python
# normalizing speaker structure
def count_turns(t):
    return len(re.findall(r"\b(caller|receiver):", t, flags=re.IGNORECASE))

df["turn_count"] = df["text"].apply(count_turns) 
```

```python
# Feature engineering
```

```python
URGENCY_WORDS = r"\b(immediately|urgent|now|verify|suspend|arrest|warrant|legal action|fine|penalty|act now|limited time|final notice)\b"
MONEY_WORDS = r"\b(gift card|wire transfer|bitcoin|crypto|bank account|social security|ssn|routing number|card number|cvv|otp|one[- ]time password)\b"
AUTHORITY_WORDS = r"\b(irs|police|fbi|government|officer|department|agent|court|federal)\b"

df["char_len"] = df["text"].str.len()
df["word_count"] = df["text"].str.split().apply(len)
df["turn_count"] = df["text"].apply(count_turns)
df["placeholder_count"] = df["text"].str.count(r"\[[A-Za-z]+\]")   
df["urgency_word_count"] = df["text"].str.lower().str.count(URGENCY_WORDS)
df["money_word_count"] = df["text"].str.lower().str.count(MONEY_WORDS)
df["authority_word_count"] = df["text"].str.lower().str.count(AUTHORITY_WORDS)
df["has_phone_number"] = df["text"].str.contains(r"\b\d{3}[-.\s]?\d{3}[-.\s]?\d{4}\b").astype(int)
df["exclaim_count"] = df["text"].str.count("!")
```
